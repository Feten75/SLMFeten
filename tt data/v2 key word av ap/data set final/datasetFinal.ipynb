```python
import json
import random

path_real = "../clean data/cleaned_data.jsonl"
path_augmented = "../data augmentation/datasetgeneree.jsonl"
path_final = "final_training_data.jsonl"

full_dataset = []
```

```python
print("Chargement des données réelles...")
try:
    with open(path_real, "r", encoding="utf-8") as f:
        for line in f:
            full_dataset.append(json.loads(line))
    print(f" {len(full_dataset)} exemples réels chargés.")
except FileNotFoundError:
    print(" Erreur : Fichier cleaned_data.jsonl non trouvé.")
```

```python
print("Chargement des données augmentées...")
try:
    count_aug = 0
    with open(path_augmented, "r", encoding="utf-8") as f:
        for line in f:
            full_dataset.append(json.loads(line))
            count_aug += 1
    print(f"{count_aug} exemples augmentés chargés.")
except FileNotFoundError:
    print(" Erreur : Fichier datasetgeneree.jsonl non trouvé.")
```

```python
# Mélanger les data pour éviter que les exemples réels et augmentés soient regroupés
# pour que l'IA ne s'habitue pas à un seul style 
print("Mélange des données en cours...")
random.seed(42)
random.shuffle(full_dataset)
```

```python
with open(path_final, "w", encoding="utf-8") as f:
    for entry in full_dataset:
        f.write(json.dumps(entry, ensure_ascii=False) + "\n")

print(f"\n FUSION TERMINÉE !")
print(f"Taille totale du dataset : {len(full_dataset)} conversations.")
print(f" Fichier créé : {path_final}")
```
